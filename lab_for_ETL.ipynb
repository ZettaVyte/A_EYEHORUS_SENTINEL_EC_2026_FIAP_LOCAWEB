## Tratamento de Valores Nulos

```python
# Preenchendo valores nulos da coluna Produto com "Nao Informado" para evitar problemas de integridade referencial no banco de dados e facilitar os métodos de feature engineering
if 'nm_produto' in dados.columns:
    dados['nm_produto'] = dados['nm_produto'].fillna('Nao Informado')
```

```python
# Preenchendo valores nulos das colunas com "Nao Informado" para evitar problemas de integridade referencial no banco de dados e facilitar os métodos de feature engineering
if 'nm_categoria' in dados.columns:
    dados['nm_categoria'] = dados['nm_categoria'].fillna('Nao Informado')

if 'nm_subcategoria' in dados.columns:
    dados['nm_subcategoria'] = dados['nm_subcategoria'].fillna('Nao Informado')
    
if 'nm_item_configuracao' in dados.columns:
    dados['nm_item_configuracao'] = dados['nm_item_configuracao'].fillna('Nao Informado')
```

```python
# Preenchendo valores nulos das colunas com valores padrão para evitar problemas de integridade referencial no banco de dados e facilitar os métodos de feature engineering
if 'cd_fechamento' in dados.columns:
    dados['cd_fechamento'] = dados['cd_fechamento'].fillna('Em andamento')
    
if 'tp_solucao' in dados.columns:
    dados['tp_solucao'] = dados['tp_solucao'].fillna('Nenhuma')
```

```python
# Substituir os valores nulos do KPI violado para verificar de acordo com o dicionário de dados se deveriam ser considerados como violados ou não
if 'fl_kpi_violado' in dados.columns:
    dados['fl_kpi_violado'] = dados['fl_kpi_violado'].fillna('VERIFICAR')
```

## Verificar regras de negócio Dicionário de Dados
